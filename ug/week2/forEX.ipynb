### 2) Print a Square of Stars
Use a nested for loop to print a n×n square of stars.

example:

n = 5
```
* * * * * 
* * * * * 
* * * * * 
* * * * * 
* * * * *
```


````python
n = 5

for i in range(n):
    line = ""
    for j in range(n):
        line += "* "
    print(line)
````

````python
n = 5

for i in range(n):
    print("* " * n)
````

### 4) Print a Pyramid of Stars
Use nested for loops to print a centered pyramid pattern of stars.

example:

n = 5
```
    * 
   * * 
  * * * 
 * * * * 
* * * * *
```

````python
n = 5

for i in range(n):
    line = ""
    for j in range(n - (i + 1)):
        line += " "
    for j in range(i + 1):
        line += "* "
    print(line)
````

````python
n = 5

for i in range(n):
    spaces = " " * (n - i - 1)
    stars = "* " * (i + 1)
    print(spaces + stars)
````

### 5) Print a Full Pyramid of Stars
Use nested for loops to print a centered full pyramid pattern of stars.

Example:

n = 5

```
        * 
      * * * 
    * * * * *  
  * * * * * * * 
* * * * * * * * *
```

````python
n = 5

for i in range(n):
    line = ""
    # Number of spaces before stars
    for j in range(n - i - 1):
        line += "  "
    # Number of stars in this row
    for j in range(2 * i + 1):
        line += "* "
    print(line)
````

````python
n = 5

for i in range(n):
    # Number of spaces before stars
    spaces = "  " * (n - i - 1)
    # Number of stars in this row
    stars = "* " * (2 * i + 1)
    print(spaces + stars)
````

### 6) Print a Christmas Tree of Stars
Use nested for loops to draw a Christmas tree made of star pyramids using n (height) and m (number of sections).

Example:

n = 5

m = 3

```
        * 
      * * * 
    * * * * * 
  * * * * * * * 
* * * * * * * * * 
      * * * 
    * * * * * 
  * * * * * * * 
* * * * * * * * * 
      * * * 
    * * * * * 
  * * * * * * * 
* * * * * * * * * 
```

````python
n = 5
m = 3

for i in range(m):
    for j in range(0 if (i == 0) else 1, n):
       # Number of spaces before stars
        spaces = "  " * (n - j - 1)
        # Number of stars in this row
        stars = "* " * (2 * j + 1)
        print(spaces + stars)
````
