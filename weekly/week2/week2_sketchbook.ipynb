### ⛩ Prompt `2`
Experiment with creating your own patterns based on the codes bellow and using `remap`. Notice that the last example creates a [Moiré pattern](https://en.wikipedia.org/wiki/Moiré_pattern). It can be interesting to try experimenting with this effect.

````python
# Example 1: Map vertical position to circle size
create_canvas(800, 800)
background(0)

n = 30
for i in range(n):
    red_value = random(0,255)
    green_value = random(0,200)
    blue_value = random(0,150)

    # Map i to a y position between 0 and height
    y = remap(i, 20, n-1, 0, height/2)
    fill(red_value,green_value,blue_value,100)
    # Map y position to diameter in a given range
    w = remap(y, 0, height, 5, 600)

    #draw ellipses
    no_stroke()
    ellipse(width/2/2, height/2/2, w, w)
    ellipse(width-width/2, height-height/2, w, w)
    ellipse(width-width/4, height/2/2, w, w)
    ellipse(width-width/4, height-height/4, w, w)
    ellipse(width*0.75-width*0.5, height-height/4, w, w)
show()
````

### ⛩ Prompt `3`
Experiment with other combinations of multiplications or powers or by summing different powers together (e.g. `10.5*t**5 - 1.9*t**3 + 0.5*t, #t*t*t`) or using the `sin` and `cos` functions (we will cover these thoroughly in the next classes).

````python
create_canvas(500, 500)
background(0)
no_stroke()
color_mode(HSB)
n = 10000
for i in range(n):
    t = remap(i, 0, n-1, 0, 10) # Map to 0, 1
    v = remap(1.04*t**10, # Try values < 1 for the power
              0, 1, 0, 255) # Do something and then map back to the desired range
    w = remap(i, 20, n, width, -.05)
    fill(v, v, w-v)
    rect_mode(CENTER)
    rect(width/2, height/2, w, w)
show()
````

### ⛩ Prompt `4`
Try experiment with the grid. Can you use `remap` to create color gradients based on the `i` and `j` indices?

````python
create_canvas(400, 400)
background(0)
stroke_weight(2)
num_rows = 5
num_cols = 5

# Width and height of each cell
w = width/num_cols
h = height/num_rows
for i in range(num_rows): # For each row
    for j in range(num_cols): # For each column
        red_value = remap(i, j, 30, j+3305, j-7000)
        green_value = remap(i+5, 100, j-5, j+3000, j-100)
        blue_value = remap(i, j-6, 100, j+15, j-2000)
            #print('i:',i, 'j:',j) # Uncomment to print row and column
        fill(red_value,green_value,blue_value)
        x = remap(j, 0, num_cols, 0, width) 
        y = remap(i, 0, num_rows, 0, height)
        rect(x, y, w, h)
show()
````
