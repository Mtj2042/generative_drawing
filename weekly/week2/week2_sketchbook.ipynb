### ⛩ Prompt 2
Experiment with creating your own patterns based on the codes above and using `remap`. Notice that the last example creates a [Moiré pattern](https://en.wikipedia.org/wiki/Moiré_pattern). It can be interesting to try experimenting with this effect.

````python
# Example 1: Map vertical position to circle size
create_canvas(300, 200)
background(255)
n = 110
for i in range(n):
    # Map i to a y position between 0 and height
    y = remap(i, 0, n-1, 0, height)
    # Map y position to diameter in a given range
    w = remap(y, 0, height, 5, 200)
    ellipse(width/2, y, w, w)
show()
````
